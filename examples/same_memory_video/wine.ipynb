# Same questions, different memory: wine
Review evidence, not a fresh experiment. Recomputes point estimates from curated numeric exports; does not rerun embeddings, Jev, timing or bootstrap intervals. Source hashes are in the eval file. Source data and notebooks for independent end-to-end reproduction remain a release prerequisite.


```python
import json
from pathlib import Path
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').exists())
data = json.loads((root / 'docs/media/same-memory/wine-eval.json').read_text())
```

```python
import math
def rmse(rows, column):
    return math.sqrt(sum((r[0] - r[column]) ** 2 for r in rows) / len(rows))
for column, name in [(1, 'venn_best'), (2, 'five_round')]:
    value = rmse(data['fresh'], column)
    assert abs(value - data['results']['fresh_3000'][name]) < 1e-10
    print(name, f'{value:.3f}')
assert abs(rmse(data['second_look'], 1) - data['results']['second_look_800']['venn_best']) < 1e-10
print('Second look:', f"{rmse(data['second_look'], 1):.3f}", 'published:', data['published_rmse'])
assert data['results']['gates'] == {'G1': True, 'G2': False}
price = data['price']
venn = price['inventory']['venn_family']
print('Whole inventory:', data['inventory_rows'])
print('Venn train + predict seconds:', venn['train_seconds'] + venn['predict_seconds'])
print('Jev direct scaled seconds:', price['jev_direct_scaled_seconds'])
print('Jev direct total USD:', price['jev_direct_total_usd'])
print('Venn API USD:', price['venn_api_usd'])
print('Separate live Jev discovery minutes:', data['jev_build']['seconds'] / 60)
print('Venn build seconds:', data['api']['build_seconds_embed_1200_and_register'])
print('Price vector cache:', price['vector_prompt'])
print('Existing vectors; embedding excluded. Jev figures cleared for publication on 2026-11-06.')
```

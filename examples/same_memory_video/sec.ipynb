```python
import json
from pathlib import Path
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').exists())
data = json.loads((root / 'docs/media/same-memory/sec-eval.json').read_text())
```

```python
for key, result in [('test', 'test_60'), ('confirmation', 'confirmation')]:
    rows = data[key]
    for column, arm in [(1, 'jev_rerun'), (2, 'venn_family_linear')]:
        correct = sum(row[0] == row[column] for row in rows)
        assert correct == data['results'][result]['arms'][arm]['group_correct']
        print(key, arm, correct, '/', len(rows))
p = data['results']['test_60']['paired_vs_jev_rerun']['venn_family_linear group']['p_one_sided']
assert p > .05
print('Primary gate FAILED; confirmation reported, not claimed.', 'p =', p)
print('Editorial example Jev confidence:', f"{data['hero']['jev_confidence']:.0%}")
```
